## 1. El método `len()` — Longitud de una cadena

````python
# 1. Longitud: ¿Cuántos caracteres tiene el nombre?
longitud = len(nom_completo)
print(f"Longitud: {longitud}")
````

### 🧪 Ejercicio

Usa `len(zika_DNA)` para contar el número de nucleótidos de un segmento de ADN del virus Zika.

![](./2.png)

````python
# Solución
zika_DNA = 'AGTTGTTGATCTGTGT'
zika_DNA_length = len(zika_DNA)

print(f'Segmento de ADN Zika: {zika_DNA}')
print(f'Número de nucleótidos: {zika_DNA_length}')
````

### 📖 Explicación

Los **nucleótidos** son moléculas orgánicas pequeñas que forman la base del ADN. El ADN está compuesto por cuatro tipos de nucleótidos, representados por las letras: **A** (adenina), **T** (timina), **G** (guanina) y **C** (citosina).

Ejemplo de un segmento real del genoma del Zika en formato GenBank:
```
agttgttgat ctgtgtgagt cagactgcga cagttcgagt ctgaagcgag agctaacaac
```

La función `len()` en Python nos permite contar cuántos caracteres (nucleótidos) tiene una cadena de texto. El genoma completo del virus Zika tiene aproximadamente **10,794 pares de bases (bp)**.

### 🧪 Ejercicio

Usa el operador `+` para concatenar varios codones que codifican la proteína GFP.

````python
# Solución
# Codones de inicio y ejemplo para la proteína GFP
S_codon = 'ATG'        # Codón de inicio (metionina)
RNA_seg = 'AGT'        # Segmento de ARN inicial

RNA_seg = RNA_seg + S_codon

print(f'Segmento de ARN concatenado: {RNA_seg}')

# Ejemplo con varios codones de GFP
codon_1 = 'ATG'
codon_2 = 'AGT'
codon_3 = 'AAA'
codon_4 = 'GGA'

secuencia_GFP = codon_1 + codon_2 + codon_3 + codon_4
print(f'Secuencia GFP ensamblada: {secuencia_GFP}')
print(f'Longitud total: {len(secuencia_GFP)} nucleótidos')
````

### 📖 Explicación

**¿Qué son los codones?**

Los codones son secuencias de **tres nucleótidos** en el ARN mensajero (ARNm) que codifican un aminoácido específico. Por ejemplo, el codón `AUG` codifica la metionina y es el codón de inicio de la traducción.

La proteína **GFP** (*Green Fluorescent Protein* — Proteína Fluorescente Verde) es ampliamente usada en biología molecular como marcador genético. Su gen está compuesto por una secuencia específica de codones.

El operador `+` en Python nos permite **concatenar** (unir) varias cadenas de texto.

## 3. El método `find()` — Buscar un codón en el D-loop del chimpancé

````python
# 3. Encontrar: ¿Dónde empieza 'Pérez'?
posicion = nom_completo.find("Doe")
print(f"El apellido empieza en el índice: {posicion}")
````

### 🧪 Ejercicio

Usa el método `find()` para localizar un codón en el D-loop del chimpancé, buscando entre las posiciones 27 y 55.

![](./3.png)

````python
# Solución
# Segmento del D-loop del chimpancé (secuencia de ejemplo)
chimp = 'ATGCTAGCTAGCTAGCTAGCTAGCTATGCGATCGATCGATCGATCGATCG'
codon = 'ATG'

# Buscar el codón entre las posiciones 27 y 55
posicion = chimp.find(codon, 27, 55)

print(f'Secuencia D-loop del chimpancé: {chimp}')
print(f'Codón buscado: {codon}')
print(f'El codón "{codon}" se encontró en la posición: {posicion}')
````

### 📖 Explicación

**¿Qué es el D-loop del chimpancé?**

Un **D-loop** (bucle de desplazamiento) es una estructura que se forma durante la reparación de roturas de doble cadena en el ADN, donde una cadena sencilla de ADN invade un dúplex homólogo, permitiendo la síntesis de ADN para recuperar información genética perdida en el sitio de rotura.

El método `find()` en Python busca una subcadena dentro de otra cadena y devuelve la posición (índice) donde se encuentra por primera vez. Puede recibir parámetros opcionales para limitar la búsqueda a un rango específico:

```python
cadena.find(subcadena, inicio, fin)
```

## 4. Segmentación (*Slicing*) — Extraer codones del D-loop humano

````python
# 4. Cortar: Extraer el primer nombre (primeras 4 letras)
apodo = nom_completo[0:4]
print(f"Apodo: {apodo}")
````

### 📖 Explicación

**¿Qué es el D-loop humano?**

El D-loop humano es una región del ADN mitocondrial que regula la replicación y transcripción del genoma mitocondrial. Es una zona de gran interés en estudios de evolución humana y medicina, ya que acumula mutaciones con mayor frecuencia que otras regiones del ADN.

El **slicing** (segmentación) en Python nos permite extraer partes de una cadena usando índices. Los índices negativos cuentan desde el final de la cadena:

```python
cadena[inicio:fin]   # extrae desde 'inicio' hasta 'fin' (sin incluirlo)
cadena[0:3]        # extrae 3 caracteres empezando desde la posición 0
```

### 🧪 Ejercicio

Usa segmentación (*slicing*) para extraer codones del D-loop humano con la notación `human[0:3]`.

````python
# Solución
# Segmento del D-loop humano (secuencia de ejemplo)
human = 'CACCATGCATGCTAGCTAGCATGCAT'

# Extraer codón usando índices negativos
codon_extraido = human[0:3]

print(f'Secuencia D-loop humano: {human}')
print(f'Longitud total: {len(human)}')
print(f'Codón extraído con [0:3]: {codon_extraido}')
````

## 5. Invertir con Reverse — Invertir el genoma del Zika

````python
# 5. Reversear: Tu nombre al revés
reverso = nom_completo[::-1]
print(reverso)
````

### 🧪 Ejercicio

Usa `zika_DNA[::-1]` para invertir un segmento del genoma del virus Zika.

````python
# Solución
zika_DNA = 'AGTTGTTGATCTGTGT'

# Invertir la secuencia
reversed_zika_DNA = zika_DNA[::-1]

print(f'Secuencia original (5→3): {zika_DNA}')
print(f'Secuencia invertida (3→5): {reversed_zika_DNA}')
print(f'Longitud: {len(zika_DNA)} nucleótidos')
````

### 📖 Explicación

**El genoma del virus Zika**

El genoma del virus Zika (ZIKV) es una molécula de ARN lineal, monocatenario y de sentido positivo, de aproximadamente **10.8 kilobases (kb)**. Esto significa que su secuencia de ARN puede traducirse directamente como si fuera ARNm.

### 🧪 Ejercicio

Usa `replace(' ', '')` para eliminar los espacios de un segmento de ADN del Zika en formato GenBank.

````python
# Solución
# Secuencia en formato GenBank (con espacios)
zika_DNA = 'agttgttgat ctgtgtgagt cagactgcga cagttcgagt ctgaagcgag agctaacaac'

print(f'Secuencia GenBank original:')
print(f'  {zika_DNA}')
print(f'  Caracteres totales: {len(zika_DNA)}')

# Eliminar espacios
zika_DNA = zika_DNA.replace(' ', '')

print(f'\nSecuencia limpia (sin espacios):')
print(f'  {zika_DNA}')
print(f'  Nucleótidos totales: {len(zika_DNA)}')
````

### 📖 Explicación

Las secuencias de ADN en formato **GenBank** (la base de datos genética más grande del mundo) a menudo incluyen espacios cada 10 nucleótidos para facilitar la lectura humana. Por ejemplo:

```
agttgttgat ctgtgtgagt cagactgcga
```
